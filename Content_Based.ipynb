```python
tfidif1 = tfidif.fit_transform(df.description)
tfidif1
```

```python
similarity_score = sigmoid_kernel(tfidif1,tfidif1)
similarity_score
```

* Here We are calculating the similrity Score with it self and all other rows using the sigmoid kernel.
* Formula to calculate the Similarity score for sigmoid kernel is:
          "similiarity_score = tanh(x1T(T->transopose)x2)+constant0.")
* We are taken the same Tfidf Vectorized data twice because we are calucating the similarity score of each row itself and all other rows.
* Means first tfidif=x1t and second tfidf=x2 so the x1t will not change it first calculate the similarity score with x2 first (itself) and save that score in the first column and next it calculate the score for next row in x2 save that score in column no2. like that it calculating the score with each row with itself and all other rows.

```python
indices = pd.Series(data=df.index,index=df.title)
indices
```

```python
def recommended_movie(similar_movie):
    if similar_movie in indices.index:
        index = np.where(similar_movie==indices.index)[0][0]
        similar = sorted(list(enumerate(similarity_score[index])),reverse=True,key=lambda x:x[1])[1:6]
        print(f'Recommended Moives of {similar_movie}')
        print('*'*30)
        for movie in similar:
            print(indices.index[movie[0]])
    else:
            print('Movie is not in the list')
```

```python
recommended_movie('Zodiac')
```
