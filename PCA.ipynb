```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.decomposition import PCA
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

* Under Scikit learn we have lots of dataset from the inbuilt data set we will do the PCA.
* The load_digits contains the images of numbers
* so the images are stroed in the pixcel format based on the hight and width.
* The maximum size of the pixcel here btw 0 to 255

```python
load_digits()
```

* So the data is added based on each image pixcel conisder a image of 8*8=64 for 64 pixcel the 64 columns are created and those data are added one by one in the columns
* so the each image and the pixcel data are added by rowise. for e.g.., for the image two the second row will be created and its data is added in the pixcel column

```python
my_data = load_digits()
print(my_data.DESCR)
# The dataset contains 1797 rows and 64 columns here
```

```python
features = pd.DataFrame(my_data.data,columns=my_data.feature_names)
# Here data will give the all features information.
features.head()
```

```python
target = pd.DataFrame(my_data.target,columns=['Target'])
target.head()
```

```python
x_train,x_test,y_train,y_test = train_test_split(features,target,train_size=0.8,random_state=100,stratify=target.Target)
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
pca = PCA(n_components=10)
# Here we are Using the first 10 PCA columnsand we will 
# the 10 column output
```

```python
x_train_new = pca.fit_transform(x_train)
x_test_new = pca.transform(x_test)
print(x_train_new.shape)
print(x_test_new.shape)
```

```python
log_model = LogisticRegression()
log_model.fit(x_train_new,y_train)
y_pred = log_model.predict(x_test_new)
accuracy_score(y_test,y_pred)
```

```python
pca.explained_variance_ratio_
```
